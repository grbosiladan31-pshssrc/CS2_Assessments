```python
highschool_ages = [11, 12, 13, 14, 15, 16, 17, 18]



#Compute Average Age
def avg_age(highschool_ages):
  average = sum(highschool_ages) / len(highschool_ages)
  print("The Average of the Highschool Ages in the list is: ", {average})





#Search Age
def search_age(highschool_ages):
  search = float(int(input("Enter the Age you want to search: "))) #Make the user input what they want to search in the list

  if search in highschool_ages:
    index_pos = highschool_ages.index(search) #Check if what the user searched for is there
    print(f'{search} is in the list in index number {index_pos}')

  else:
    print(f'Sorry, but {search} is not in the list')






#Update the ages
def update_age(highschool_ages):
  new_age = int(input("Please enter the Age you want to change: "))
  index_age = int(input("Please enter the index you want updated: "))
  highschool_ages[index_age] = new_age
  print(f'YAY!, index {index_age} is successfuly updated to {new_age}')





#Display ALL the ages
def display_age(ages_list):
  print("Current Ages: ", ages_list)



#Ask the User What Operation they would like to do

while True:
  operation = int(input("What would you like to do? 1-See average age / 2-Search Age / 3-Update Age / 4-Display Updated Ages / 5-Exit"))

  if operation == 1:
    avg_age(highschool_ages)

  elif operation == 2:
    search_age(highschool_ages)

  elif operation == 3:
    update_age(highschool_ages)

  elif operation == 4:
    print(highschool_ages)

  elif operation == 5:
    print("Exiting the App...")
    break

  else:
    print("Wrong, Input only numbers 1-5")



```
